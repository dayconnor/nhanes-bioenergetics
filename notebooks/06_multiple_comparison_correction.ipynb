```python
from statsmodels.stats.multitest import multipletests
```

```python
hypotheses = ['H1 (triglycerides)', 'H2 (free T3 coefficient)', 'H3 (CRP)', 'H4 (cholesterol)']
# Primary-test p-values, from notebooks 02 (model a), 03 (ft3_z), 04 (log CRP),
# 05 (primary). H2's pre-specified AUC comparison has no p-value, so the free T3
# coefficient stands in for H2 in this family.
p_values = [7.36e-07, 0.000133, 0.00311, 0.0938]

reject_fdr, p_values_fdr, _, _ = multipletests(p_values, alpha=0.05, method='fdr_bh')
reject_bonf, p_values_bonf, _, _ = multipletests(p_values, alpha=0.05, method='bonferroni')

for h, p, rf, pf, rb, pb in zip(hypotheses, p_values, reject_fdr, p_values_fdr, reject_bonf, p_values_bonf):
    print(f'{h}: raw p={p:.3g} | FDR-adj p={pf:.3g} (sig={rf}) | Bonferroni-adj p={pb:.3g} (sig={rb})')
```
